```python

import pandas as pd

# Load the dataset
df = pd.read_csv("/content/BBMP_Gemma.csv")

# Select the required columns and remove missing values
data = df[["Staff Remarks", "Grievance Status"]].dropna().copy()

# Clean the text and labels
data["text"] = data["Staff Remarks"].astype(str).str.strip().str.lower()
data["label"] = data["Grievance Status"].astype(str).str.strip()

# Remove empty remarks
data = data[data["text"] != ""]

# Remove duplicate text-label combinations
data = data.drop_duplicates(subset=["text", "label"])

# Keep only remarks associated with one status
label_counts = data.groupby("text")["label"].nunique()
valid_texts = label_counts[label_counts == 1].index
clean_data = data[data["text"].isin(valid_texts)].copy()

# Keep status classes with at least 10 records
counts = clean_data["label"].value_counts()
valid_labels = counts[counts >= 10].index
clean_data = clean_data[clean_data["label"].isin(valid_labels)]

print("Original dataset size:", len(df))
print("Usable records:", len(clean_data))
print("\nStatus counts:")
print(clean_data["label"].value_counts())
```

```python

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report
import pandas as pd

# Prepare input and target
X = clean_data["text"]
y = clean_data["label"]

# Split data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))

# Build the machine-learning pipeline
model = Pipeline([
    ("tfidf", TfidfVectorizer(
        ngram_range=(1, 2),
        max_features=10000
    )),
    ("classifier", LogisticRegression(
        max_iter=1000,
        class_weight="balanced"
    ))
])

# Train the model
model.fit(X_train, y_train)

# Evaluate predictions
predictions = model.predict(X_test)

print("\nModel Accuracy:",
      round(accuracy_score(y_test, predictions), 4))

print("\nClassification Report:")
print(classification_report(
    y_test,
    predictions,
    zero_division=0
))
```

```python

# STEP 2: Train the machine-learning model

from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Prepare the input and output
X = clean_data["text"]
y = clean_data["label"]

# Split data: 80% for training and 20% for testing
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training records:", len(X_train))
print("Testing records:", len(X_test))

# Create the model
model = Pipeline([
    ("tfidf", TfidfVectorizer(
        ngram_range=(1, 2),
        max_features=10000
    )),
    ("classifier", LogisticRegression(
        max_iter=1000,
        class_weight="balanced"
    ))
])

# Train the model
model.fit(X_train, y_train)

print("\nModel training completed successfully!")

# Test the model
predictions = model.predict(X_test)

# Display performance
print("\nAccuracy:",
      round(accuracy_score(y_test, predictions), 4))

print("\nClassification Report:")
print(classification_report(
    y_test,
    predictions,
    zero_division=0
))
```
